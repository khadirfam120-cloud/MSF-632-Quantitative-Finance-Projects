# FNAN 632 - Project Part 1

For this project I used IVV monthly price data and calculated the returns and basic statistics for Jan. 2016 through Dec. 2025.

## Setup

```python
import yfinance as yf
import pandas as pd
import numpy as np
```

## 1. Download Monthly IVV Prices

I used the same `yfinance` method as the class example and started the data in May 2000.

```python
ivv = yf.Ticker("IVV")
prices = ivv.history(interval="1mo", auto_adjust=True, start="2000-05-01")
prices.head()
```

I also checked the end of the data to make sure it downloaded correctly.

```python
prices.tail()
```

## 2. Compute Monthly Percent Returns

I used percent change on the adjusted closing prices. I calculated the returns first and then selected 2016-2025 so January 2016 can use the December 2015 price.

```python
prices["Return"] = prices["Close"].pct_change()

returns = prices.loc[
    (prices.index >= "2016-01-01") & (prices.index < "2026-01-01"),
    "Return"
].dropna()

print("Number of monthly returns:", len(returns))
returns.head()
```

There should be 120 monthly returns because the sample covers 10 years.

## 3. Arithmetic Monthly Mean Return

```python
arithmetic_monthly_mean = returns.mean()
print("Arithmetic monthly mean return:", arithmetic_monthly_mean)
print("Percent:", round(arithmetic_monthly_mean * 100, 4), "%")
```

This is the average monthly return over the sample.

## 4. Value of a $1 Investment

I compounded all of the monthly returns to find the ending value of $1.

```python
ending_value = (1 + returns).prod()
print("Ending value of $1: $", round(ending_value, 4))
```

So this is what $1 invested before January 2016 would be worth at the end of 2025.

## 5. Geometric Mean Return

I used the ending value from Question 4 to calculate the geometric monthly mean.

```python
geometric_monthly_mean = ending_value ** (1 / len(returns)) - 1
print("Geometric monthly mean:", geometric_monthly_mean)
print("Percent:", round(geometric_monthly_mean * 100, 4), "%")
```

The geometric mean accounts for compounding over the full period.

## 6. Median

```python
median_return = returns.median()
print("Median:", median_return)
print("Percent:", round(median_return * 100, 4), "%")
```

## 7. Quartiles

```python
quartiles = returns.quantile([0.25, 0.50, 0.75])
quartiles
```

P50 is the same as the median from Question 6.

## 8. Quintiles

```python
quintiles = returns.quantile([0.20, 0.40, 0.60, 0.80])
quintiles
```

## 9. 10th and 90th Percentiles

```python
p10 = returns.quantile(0.10)
p90 = returns.quantile(0.90)

print("P10:", p10)
print("P90:", p90)
```

## 10. Range and MAD

The range is the maximum return minus the minimum return. For MAD, I used mean absolute deviation from the mean.

```python
sample_range = returns.max() - returns.min()
mad = np.mean(np.abs(returns - returns.mean()))

print("Range:", sample_range)
print("Range (%):", round(sample_range * 100, 4), "%")
print("MAD:", mad)
print("MAD (%):", round(mad * 100, 4), "%")
```

These are two different ways to describe the spread of the monthly returns.

# Annualizing the Returns

## 11. Annual Growth Rate Using the Arithmetic Mean

Since the question says this is similar to APR, I multiplied the monthly arithmetic mean by 12.

```python
annual_arithmetic_rate = arithmetic_monthly_mean * 12
print("Annual arithmetic rate:", round(annual_arithmetic_rate * 100, 4), "%")
```

This annualizes the monthly average without compounding it.

## 12. Annualized Geometric Return

I used the ending value from Question 4 and the 10-year formula from the assignment.

```python
annual_geometric_rate = ending_value ** (1 / 10) - 1
print("Annualized geometric return:", round(annual_geometric_rate * 100, 4), "%")
```

This is the constant annual rate that would give the same ending value over the 10 years.

## 13. Which Metric Would I Recommend?

I would use the annualized geometric return. The arithmetic return gives the average, but the geometric return takes compounding into account. Since an investment grows from one period to the next, I think the geometric return gives a better picture of the ETF's actual growth over the 10 years.
